```python
import pandas as pd





```

```python
df = pd.read_excel("plan_de_compras_2025.xlsx")
df.head()
```

```python
# Extraer los últimos 2 dígitos del código de proyecto (el año)

df["Año Proyecto"] = df["ID Proyecto"].astype(str).str[-2:]

 

# Filtrar solo los proyectos del año 2025 (código termina en "25")

df_2025 = df[df["Año Proyecto"] == "25"]

 

print(f"Total de registros: {len(df)}")

print(f"Registros año 2025: {len(df_2025)}")

 

df_2025.head(20)
```
