```python
import pandas as pd

df = pd.read_excel("../plan_de_compras_2025.xlsx")

print("Archivo cargado correctamente")
print("Filas y columnas:", df.shape)
```

```python
resumen_tipo_proyecto = df.groupby("Tipo Proyecto").agg(
    Cantidad_Registros=("ID Proyecto", "count"),
    Monto_Total=("Monto Total Ítem Año 2025", "sum")
).reset_index()

resumen_tipo_proyecto
```

```python
resumen_estado = df.groupby("Estado Proyecto").agg(
    Cantidad_Registros=("ID Proyecto", "count"),
    Monto_Total=("Monto Total Ítem Año 2025", "sum")
).reset_index()

resumen_estado
```

```python
resumen_unidad = df.groupby("Unidad de Compra").agg(
    Cantidad_Registros=("ID Proyecto", "count"),
    Monto_Total=("Monto Total Ítem Año 2025", "sum")
).reset_index()

resumen_unidad = resumen_unidad.sort_values(
    "Monto_Total",
    ascending=False
)

resumen_unidad
```

```python
resumen_tipo_compra = df.groupby("Tipo compra").agg(
    Cantidad_Registros=("ID Proyecto", "count"),
    Monto_Total=("Monto Total Ítem Año 2025", "sum")
).reset_index()

resumen_tipo_compra = resumen_tipo_compra.sort_values(
    "Monto_Total",
    ascending=False
)

resumen_tipo_compra
```
