```python
import pandas as pd
```

```python
df = pd.read_excel("../plan_de_compras_2025.xlsx")
```

```python
# Buscar columnas relacionadas con teléfono
[col for col in df.columns if "tel" in col.lower()]
```

```python
display(df["Teléfono responsable"].head(20))
```

```python
print("Nulos:", df["Teléfono responsable"].isna().sum())
print("Valores únicos:", df["Teléfono responsable"].nunique())
```

```python
columna_telefono = "Teléfono responsable"

df["telefono_limpio"] = (
    df[columna_telefono]
    .astype("string")
    .str.replace(r"\D+", "", regex=True)  # deja solo dígitos   
    .replace(["", "0"], pd.NA)
)

display(df[[columna_telefono, "telefono_limpio"]].head(20))
print("Nulos en teléfono limpio:", df["telefono_limpio"].isna().sum())
```

```python
df["largo_telefono"] = df["telefono_limpio"].str.len()

display(
    df["largo_telefono"]
    .value_counts(dropna=False)
    .sort_index()
    .to_frame("cantidad")
)
```

```python
filas_antes = len(df)

df = df.dropna(subset=["telefono_limpio"]).copy()

print("Filas eliminadas:", filas_antes - len(df))
print("Nulos restantes en teléfono:", df["telefono_limpio"].isna().sum())
```
